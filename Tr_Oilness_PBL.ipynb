```python
import os

# Sesuaikan dengan path folder dataset Anda
# Struktur klasifikasi biasanya terbagi menjadi folder: train, valid/val, dan test
dataset_dir = '/content/Oily-1'

print("Jumlah gambar pada data Training:")
for nama_kelas in os.listdir(dataset_dir):
    path_kelas = os.path.join(dataset_dir, nama_kelas)

    # Memastikan yang dihitung hanya folder kelas
    if os.path.isdir(path_kelas):
        jumlah_file = len(os.listdir(path_kelas))
        print(f"- Kelas '{nama_kelas}': {jumlah_file} gambar")
```

```python
import os

dataset_dir = '/content/Oily-1'
splits = ['train', 'valid', 'test']

for split in splits:
    split_path = os.path.join(dataset_dir, split)
    if os.path.exists(split_path):
        print(f"\n=== Distribusi Kelas pada Data {split.upper()} ===")
        total_images = 0
        for nama_kelas in sorted(os.listdir(split_path)):
            path_kelas = os.path.join(split_path, nama_kelas)
            if os.path.isdir(path_kelas):
                # Menghitung file gambar saja (jpg, jpeg, png)
                jumlah_file = len([f for f in os.listdir(path_kelas) if f.lower().endswith(('.png', '.jpg', '.jpeg'))])
                print(f"- Kelas '{nama_kelas}': {jumlah_file} gambar")
                total_images += jumlah_file
        print(f"Total gambar pada {split}: {total_images}")
    else:
        print(f"Direktori {split} tidak ditemukan.")
```
